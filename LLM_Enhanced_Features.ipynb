## Part 2: Setup Database

First, we need to set up the SQLite database (same as before).

```python
# Run database setup from the main notebook
# This creates the support.db with customers and tickets

import sqlite3
from datetime import datetime

class DatabaseSetup:
    """SQLite database setup for customer support system."""
    
    def __init__(self, db_path: str = "support.db"):
        self.db_path = db_path
        self.conn = None
        self.cursor = None

    def connect(self):
        self.conn = sqlite3.connect(self.db_path)
        self.conn.execute("PRAGMA foreign_keys = ON")
        self.cursor = self.conn.cursor()

    def create_tables(self):
        # Create customers table
        self.cursor.execute("""
            CREATE TABLE IF NOT EXISTS customers (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                name TEXT NOT NULL,
                email TEXT,
                phone TEXT,
                status TEXT NOT NULL DEFAULT 'active',
                created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
                updated_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
            )
        """)

        # Create tickets table
        self.cursor.execute("""
            CREATE TABLE IF NOT EXISTS tickets (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                customer_id INTEGER NOT NULL,
                issue TEXT NOT NULL,
                status TEXT NOT NULL DEFAULT 'open',
                priority TEXT NOT NULL DEFAULT 'medium',
                created_at DATETIME DEFAULT CURRENT_TIMESTAMP,
                FOREIGN KEY (customer_id) REFERENCES customers(id)
            )
        """)
        self.conn.commit()

    def insert_sample_data(self):
        customers = [
            ("John Doe", "john.doe@example.com", "+1-555-0101", "active"),
            ("Jane Smith", "jane.smith@example.com", "+1-555-0102", "active"),
            ("Alice Williams", "alice.w@techcorp.com", "+1-555-0104", "active"),
            ("Charlie Brown", "charlie.brown@email.com", "+1-555-0105", "active"),
        ]
        self.cursor.executemany("""
            INSERT OR IGNORE INTO customers (name, email, phone, status)
            VALUES (?, ?, ?, ?)
        """, customers)

        tickets = [
            (1, "Cannot login to account", "open", "high"),
            (2, "Password reset not working", "in_progress", "medium"),
            (3, "Email notifications not received", "open", "medium"),
        ]
        self.cursor.executemany("""
            INSERT OR IGNORE INTO tickets (customer_id, issue, status, priority)
            VALUES (?, ?, ?, ?)
        """, tickets)
        self.conn.commit()

    def close(self):
        if self.conn:
            self.conn.close()

# Initialize database
db = DatabaseSetup()
db.connect()
db.create_tables()
db.insert_sample_data()
db.close()

print("✓ Database ready!")
```

## Part 7: Cost Analysis

Understanding the cost of using LLM-powered agents.

```python
# Cost calculator
def estimate_cost(num_queries, model="gpt-4o-mini"):
    """Estimate cost for running queries."""
    
    costs = {
        "gpt-4o-mini": {"input": 0.15, "output": 0.60},  # per 1M tokens
        "gpt-4o": {"input": 2.50, "output": 10.00},
        "gpt-4-turbo": {"input": 10.00, "output": 30.00}
    }
    
    # Estimate ~150 input tokens, ~100 output tokens per query
    input_tokens = num_queries * 150
    output_tokens = num_queries * 100
    
    model_costs = costs[model]
    input_cost = (input_tokens / 1_000_000) * model_costs["input"]
    output_cost = (output_tokens / 1_000_000) * model_costs["output"]
    total_cost = input_cost + output_cost
    
    return {
        "queries": num_queries,
        "model": model,
        "input_tokens": input_tokens,
        "output_tokens": output_tokens,
        "total_cost": round(total_cost, 4),
        "cost_per_query": round(total_cost / num_queries, 6) if num_queries > 0 else 0
    }

# Example calculations
print("Cost Estimates:")
print("="*80)

for queries in [10, 100, 1000, 10000]:
    cost = estimate_cost(queries, "gpt-4o-mini")
    print(f"{queries:5d} queries: ${cost['total_cost']:8.4f}  (${cost['cost_per_query']:.6f} per query)")

print("\n💡 For this notebook's 5 test scenarios: < $0.01")
print("💡 For 1000 queries/day: ~$0.03/day = $0.90/month")
```

## Conclusion: LLM vs Rule-Based

### When to Use LLM-Powered:
✅ Production customer-facing systems  
✅ Need to handle diverse, unpredictable queries  
✅ Quality of response is critical  
✅ Have budget for API costs ($0.03-0.10 per 100 queries)  
✅ Want minimal maintenance (no pattern updates)  

### When to Use Rule-Based:
✅ Internal tools with predictable queries  
✅ Prototype/demo with no API access  
✅ Zero-cost requirement  
✅ Need maximum speed (<100ms responses)  
✅ Limited query vocabulary (controlled input)  

### Best Practice:
🎯 **Hybrid Approach**: Use LLM for complex queries, fall back to rules for simple ones

### Key Takeaways:

1. **LLM adds intelligence** - Understands nuance, context, and intent
2. **Cost is minimal** - ~$0.0001 per query with gpt-4o-mini
3. **Fallback is built-in** - System works without LLM
4. **Easy to switch** - Just toggle `use_llm=True/False`
